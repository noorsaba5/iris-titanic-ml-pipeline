### A unifying framework

- Most machine learning can be framed in a specific way. 
- Framing novel techniques like this will be helpful for YOUR understanding of what these things do, and help to demistify``ML"
- Will get introduced to many algorithms/techniques but you will realise that they are all very related if you can learn how to frame them like this


So what are the components:

- The underlying model, how you are trying to model the data, your $f(x)$. Often assumes a given structure/probability distribution. Models are parametrised- that is they hope to learn. This is what all ML does, it is finding the best parameters for an assumed model. 
    - In a linear regression, the model is a straight line. We are supposing a straight line underpins the data. 
    - In a neural network, the model is a series of linear transformations and non-linearities. We are supposing that a series of linear transformations and non-linearities can model the data from the input to the output. 

- What is learnt, i.e the parameters of the model.
    - In a linear regression, the parameters are the slope and intercept of the line.
    - In a neural network, the parameters are the weights and biases of the network. 

- How the parameters are learnt. This is through a loss function. The assumed best parameters are those that minimise the loss function. 
    - In linear regression, the loss function is the sum of the squared errors. 
    - In a neural network, the loss function can be many things, it depends what the output is. It could also be sum of squared errors.

- How practically the minima of the loss function (i.e the parameters) are found. E.g an optimisation algorithm.
    - In a linear regression, there is an analytical solution to find the minima of the loss function.
    - In a neural network, we use stochastic gradient descent. (You could use gradient descent with linear regression)
    - Sometimes this is heuristic (i.e not guaranteed to find the minima, use some sort of algorithm to ``do better")

All ML is essentially an optimisation problem, computers ``learn" how to find the best parameters for a given model by minimising a loss function.

### A note on train/test split:

- A best test set ideally would be separate from your original data. I.e an unseen dataset.
- Be really careful about choosing a test set from your trianing data. 
    - Data leakage- including dependent data when you have assumed it's independent. E,g many observations on many patients, including different observations in both the training and test set that are from the same patient.
    - If you do a random split, but some of your test data is very similar to your training data, you may get a false sense of how well your model is doing. 
    - These are common problems in ML and leads to lots of misreporting of ``good results".


### The make it go away ethos:

- If you find a result that is too good to be true, it probably is. 
- You should try everything in your power to make it go away, and if you can't, then you can start to believe it.
- So much bad data science occurs because not enough reflection on the ways that the data/modelling could be wrong.
