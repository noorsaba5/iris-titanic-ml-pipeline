### Loops

Loop is used to iterate over a set of statements multiple times.

Note: Avoid infinite loops

#### 1. for loop

```python
# range(start, stop, step)
# start - starting point - 0
# stop - ending point [exclusive] - we have to specify
# step - interval - 1

# Note: 
# if we are moving in positive direction, the ending point is end - 1
# if we are moving in negative direction, the ending point is end + 1

list(range(10, 0, -1))
```

```python
table = 5

for x in range(1, 11): 
    print(f"{table} x {x} = {table*x}")
```

```python
# Print the table in the reverse order

for x in range(10, 0, -1): 
    print(f"{table} x {x} = {table*x}")
```

#### 2. while loop

Useful when we don't till what point we have to iterate.

Syntax:

while condition/s:
    statement/s
    exit condition -> Without this condition, this while loop will be an infinite loop.

```python
table = 5

x = 1
while x < 11:
    print(f"{table} x {x} = {table*x}")
    x = x + 1 # Exit condition
```

```python
# Identify the even vs odd numbers from 1 to 100.

for item in range(1,101):
    if item % 2 == 0:
        print(f'{item}: even')
    else:
        print(f'{item}: odd')
```

```python
# Given a number, identify the number of digits it hold.
# 103 - 3
# 1324124 - 7

# len(str(103)) -> not allowed
# Conversion to string isn't allowed
```

```python
item = 2
cnt = 0
while item > 0:
    item = item // 10 # Integer division
    cnt+=1 # cnt = cnt + 1
print(cnt)
```

```python
i = input("Enter a number: ")
number = int(i)
number_of_digits = 0
while number > 9:
    number = number / 10
    print(number)
    number_of_digits += 1
print(f"The number {i} has {number_of_digits + 1} digits.")
```
