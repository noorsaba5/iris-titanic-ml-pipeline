# 2 states - 
# a. Select State - Use Esc in a cell to come to select state from edit state. Used to execute shortcuts
    # ESC + A - Define a new cell above the current cell
    # ESC + B - Define a new cell below the current cell
    # ESC + DD - Deleting a cell
    # ESC + C - Copy
    # ESC + V - Paste
    # ESC + X - Cut

# b. Edit State (Explore)
    # Shift + Enter
    # Ctrl + Enter

# 3 modes
# a. Code Mode (ESC + Y)
    # We can write syntatically correct python code.
# b. Markdown mode (ESC + M)
    # Where we can define the cell as a markdown to define text, tables, embedding images, writing html, writing latex (formatted text info)
# c. RawNBConvert Mode (ESC + R)
    # This mode is just like plain text without formatting
