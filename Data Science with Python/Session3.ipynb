### Conditional statement - Loops

Loop is used to iterate over a set of statements multiple times.

Note: Avoid infinite loops

```python
# Given a string, identify if there are any duplicate characters that exist.
# Ex - 'abcatfghb' -> yes
# Ex - 'abcdrthgyu' -> No
s1 = 'abcdrthgyu' # List/array of characters
# Takes O(n^2)
flag = False
for idx in range(len(s1)): # O(n)
    print(idx, s1[idx], s1[idx+1:])
    if s1[idx] in s1[idx+1:]: # In string, we can use the concept of slicing/dicing
        flag = True
        break

if flag:
    print('yes')
else:
    print('No')
```

```python
# optimization can happen in 2 different forms:
# time optimization -> Time complexity
# space optimization ->  Space complexity

# O(1), O(log n), O(n), O(nlogn), O(n^2), O(n^3), O(n!), O(x^n) => Worst case scenario

# We give priority to time complexity and there is always a trade off in between time and space complexity.
```

```python
# Given 2 strings, s1 and s2, identify if these strings have duplicate characters.
# s1 = 'abc' & s2 = 'bfg' => yes
# s1 = 'abc' & s2 = 'xyz' => no

s1 = 'abcdef' # m characters
s2 = 'xyz' # n characters

flag = False
for idx in range(len(s1)):
    if s1[idx] in s2: # In string, we can use the concept of slicing/dicing
        flag = True
        break

if flag:
    print('yes')
else:
    print('No')

# What's the time complexity?
# O(m*n)
```

```python
# Display in output - value + data type of the value
# (2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7)
tup = (2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7)
for item in tup:
    if type(item) == int:
        print(item, 'int')
    elif type(item) == float:
        print(item, 'float')
    elif type(item) == str:
        print(item, 'string')
    elif type(item) == bool:
        print(item, 'boolean')
    else:
        print(item, 'Incorrect value')
```
