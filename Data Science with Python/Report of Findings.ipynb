```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# AI/ML Bootcamp Exercise: Data Analysis and Handling\n",
    "This notebook demonstrates data loading, cleaning, and analysis on a sample customer dataset."
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Step 1: Import Libraries"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "import pandas as pd\n",
    "import numpy as np\n",
    "import matplotlib.pyplot as plt\n",
    "import seaborn as sns\n",
    "%matplotlib inline"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Step 2: Create Sample Dataset (No CSV Needed)"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "# Sample dataset with duplicates, missing values, and potential outliers\n",
    "data = {\n",
    "    'CustomerID': [1, 2, 3, 4, 5, 5],\n",
    "    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Eve'],\n",
    "    'Age': [25, 30, 35, np.nan, 28, 28],\n",
    "    'Gender': ['F', 'M', 'M', 'M', np.nan, 'F'],\n",
    "    'Annual Income': [50000, 60000, 70000, 80000, 90000, 90000],\n",
    "    'Spending Score': [60, 70, 80, 90, 100, 100]\n",
    "}\n",
    "\n",
    "df = pd.DataFrame(data)\n",
    "print(\"Sample dataset:\")\n",
    "print(df)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Step 3: Identify Duplicate Rows"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "# Detect duplicates\n",
    "duplicates = df[df.duplicated()]\n",
    "print(f\"Number of duplicate rows: {duplicates.shape[0]}\")\n",
    "print(duplicates)\n",
    "\n",
    "# Remove duplicates\n",
    "df = df.drop_duplicates()"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Step 4: Handle Missing Values"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "# Check missing values\n",
    "print(df.isnull().sum())\n",
    "\n",
    "# Fill numerical missing values with median\n",
    "for col in ['Age', 'Annual Income', 'Spending Score']:\n",
    "    df[col].fillna(df[col].median(), inplace=True)\n",
    "\n",
    "# Fill categorical missing values with mode\n",
    "df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)\n",
    "\n",
    "print(\"\\nMissing values after handling:\")\n",
    "print(df.isnull().sum())"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Step 5: Detect and Handle Outliers"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "# Function to detect outliers using IQR\n",
    "def detect_outliers(column):\n",
    "    Q1 = column.quantile(0.25)\n",
    "    Q3 = column.quantile(0.75)\n",
    "    IQR = Q3 - Q1\n",
    "    lower_bound = Q1 - 1.5 * IQR\n",
    "    upper_bound = Q3 + 1.5 * IQR\n",
    "    outliers = column[(column < lower_bound) | (column > upper_bound)]\n",
    "    return outliers\n",
    "\n",
    "# Detect outliers\n",
    "for col in ['Age', 'Annual Income', 'Spending Score']:\n",
    "    outliers = detect_outliers(df[col])\n",
    "    print(f\"{col} outliers:\\n{outliers}\\n\")\n",
    "\n",
    "# Visualize outliers\n",
    "plt.figure(figsize=(12,6))\n",
    "sns.boxplot(data=df[['Age', 'Annual Income', 'Spending Score']])\n",
    "plt.title('Boxplot for Outlier Detection')\n",
    "plt.show()"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### Handle Outliers using Winsorization"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "# Cap outliers at upper and lower bounds\n",
    "for col in ['Age', 'Annual Income', 'Spending Score']:\n",
    "    Q1 = df[col].quantile(0.25)\n",
    "    Q3 = df[col].quantile(0.75)\n",
    "    IQR = Q3 - Q1\n",
    "    lower = Q1 - 1.5*IQR\n",
    "    upper = Q3 + 1.5*IQR\n",
    "    df[col] = np.where(df[col] < lower, lower, df[col])\n",
    "    df[col] = np.where(df[col] > upper, upper, df[col])"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Step 6: Final Cleaned Dataset"
   ]
  },
  {
   "cell_type": "code",
   "metadata": {},
   "source": [
    "print(\"Cleaned dataset:\")\n",
    "print(df)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## Summary\n",
    "- Sample dataset created within the notebook.\n",
    "- Duplicate rows identified and removed.\n",
    "- Missing values handled using median/mode.\n",
    "- Outliers detected using IQR and handled using Winsorization.\n",
    "- Visualizations provided for outlier detection.\n",
    "\n",
    "This notebook is fully self-contained and ready to submit."
   ]
  }
 ],
 "metadata": {
  "kernelspec": {
   "display_name": "Python 3",
   "language": "python",
   "name": "python3"
  },
  "language_info": {
   "name": "python",
   "version": "3.12"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 5
}
```
