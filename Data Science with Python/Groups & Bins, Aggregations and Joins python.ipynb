# 2. Types of Binning

There are 3 major ways to create bins:

| Method     | Meaning                            | Good When…                                                     |
| ---------- | ---------------------------------- | -------------------------------------------------------------- |
| `pd.cut`   | Equal-width bins                   | You want fixed ranges                                          |
| `pd.qcut`  | Equal-frequency bins               | You want each bin to contain roughly the same number of people |
| `np.where` | Rule-based / domain knowledge bins | You want custom categories                                     |


# Binning with pd.cut() ----- (Equal-Width Bins)
PURPOSE:
Automatically divided data into bins of equal numeric width.
# Example:
Let's devide AGE into 4 equal sized bins.


```python
import pandas as pd 
#sample data 
df = pd.DataFrame({ 'age' : [11,15,22,27,35,46,52,61,74,89]}) 
#create 4 equal-width bins
df['age_cut'] = pd.cut(df['age'], bins=4)
df
```

# Binning with pd.cut() + Labels
you can add meaningful labels.

```python
bins = [10,20,30,50,70,100]
labels = [ 'teen', 'adult', 'young_mid', 'mid_age', 'senior']

df['age_group']=pd.cut(df['age'], bins=bins, labels=labels)

df
```

# Binning with pd.qcut() ----- (Equal- Frequency Bins)
Purpose:
Divides data so that each bin has almost the same number of observation.
if you have 1000 people and create 4 bins ---- each bin has ~250 people.
# Example:


```python
df ['age_qcut'] = pd.qcut(df['age'],q=4)
df
```
