```python
# Lambda Functions - Anonymous functions (for simple operations/functions)

# syntax: lambda x, y,...: operation(s)

add_op = lambda x, y: x + y

add_op(2,3) # Here the variale is acting as a function. This line defeats the whole purpose of lambda function
```

```python
# map, filter, reduce
# Addition operation
list(map(lambda x, y: x + y, [1,2], [5,6])) # Element wise operation
```

```python
# Multiplication

list(map(lambda x, y: x * y, [1,2,3], [5,6])) # Discarding the extra element.
```

```python
# Check out if the given element in a list is even or not using lambda function

lst = [2,5,4,77,44,33]

list(map(lambda x: x%2==0, [2,5,4,77,44,33]))
```

```python
# Filter out the elements in a list that odd using lambda function

list(filter(lambda x: x%2==0, [2,5,4,77,44,33]))
```

```python
sample = ['GTA','GGG','CAC']
# Write a UDF to read the file by name and return the content
# saved in the file

def read_dna(dna_file):
    """
    This function reads the file
    input: file name as a string
    output: file data as a string
    """
    dna_data = ''

    with open(dna_file, 'r') as f:
        for line in f:
            dna_data += line # Arithemetic assignment operator
    
    return dna_data

suspect1 = read_dna('suspect1.txt')
suspect2 = read_dna('suspect2.txt')
suspect3 = read_dna('suspect3.txt')
```

```python
def dna_codons(dna):
    """
    This function will create a list of codons from given string
    input: dna sample as a string
    output: list of codons (string)
    """
    codon = []
    for i in range(0, len(dna), 3):
        if i+3 > len(dna):
            break
        codon.append(dna[i:i+3])
    return codon

codon1 = dna_codons(suspect1)
codon2 = dna_codons(suspect2)
codon3 = dna_codons(suspect3)
```

```python
def match_dna(dna):
    """
    It matches the given codon to the suspect's DNA
    input: codon info as a string
    output: total matches as an integer
    """
    matches = 0
    for item in dna:
        if item in sample:
            matches+=1
    return matches

match_dna(codon1)
```
