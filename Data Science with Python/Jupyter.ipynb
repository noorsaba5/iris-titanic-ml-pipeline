```python
# 💰 Personal Expense Tracker – Python Project
# Jupyter Notebook Version (VS Code Compatible)

print("📌 Personal Expense Tracker Loaded Successfully!")
```

```python
# ---------------- Initialize Data ----------------

expenses = []  # List storing all expenses
categories = {"Food", "Transport", "Entertainment", "Utilities"}  # Set of categories

print("✅ Data initialized.")
```

```python
# ---------------- Functions ----------------

def add_expense(date, category, amount, description):
    if category not in categories:
        print(f"⚠️ Category '{category}' not recognized. Adding it...")
        categories.add(category)

    expenses.append({
        "date": date,
        "category": category,
        "amount": amount,
        "description": description
    })
    print(f"🟢 Added: {description} | {category} | ${amount}")


def view_expenses():
    print("\n📄 --- All Expenses ---")
    if not expenses:
        print("No expenses recorded yet.")
        return

    for i, e in enumerate(expenses, start=1):
        print(f"{i}. {e['date']} | {e['category']} | ${e['amount']} | {e['description']}")


def update_expense(index, key, value):
    if 0 <= index < len(expenses):
        expenses[index][key] = value
        print(f"✏️ Updated Expense {index+1}: {key} = {value}")
    else:
        print("❌ Invalid index.")


def delete_expense(index):
    if 0 <= index < len(expenses):
        removed = expenses.pop(index)
        print(f"🗑️ Removed: {removed}")
    else:
        print("❌ Invalid index.")


def total_by_category():
    summary = {}
    for e in expenses:
        summary[e["category"]] = summary.get(e["category"], 0) + e["amount"]

    print("\n📊 --- Total by Category ---")
    for cat, amt in summary.items():
        print(f"{cat}: ${amt}")

    return summary
```

```python
# ---------------- Demo ----------------

add_expense("2025-11-19", "Food", 15.5, "Lunch")
add_expense("2025-11-19", "Transport", 7.0, "Bus fare")
add_expense("2025-11-19", "Entertainment", 12.0, "Movie ticket")

print("\n📌 Viewing all expenses:")
view_expenses()

print("\n📌 Updating expense 2:")
update_expense(1, "amount", 8.0)

print("\n📌 Deleting expense 3:")
delete_expense(2)

print("\n📌 Viewing final list:")
view_expenses()

print("\n📌 Category summary:")
total_by_category()
```
