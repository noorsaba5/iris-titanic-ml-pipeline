# Groups & Bins

When working with continuous numeric variables such as Age, we often categorize them into groups or bins. This enables aggregated analysis, segmentation, and easier interpretation.

Your age range spans from 11 to 90, giving 80 unique integer values.

# Creating Groups Based on Domain Knowledge

Domain knowledge–driven grouping is subjective and contextual. It reflects human understanding rather than purely statistical boundaries.
Example categories:

Age Range	Category
11–19	Teenagers
20–30	Adults
30–50	Mid Age
>50	Old

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Age': [15, 22, 37, 51, 70]})

df['Age_Group'] = np.where(df['Age'] <= 19, 'Teenagers',
                   np.where(df['Age'] <= 30, 'Adult',
                   np.where(df['Age'] <= 50, 'Mid age', 'Old')))

print(df)
```

# Creating Equal-Spaced Bins

Equal-spaced bins divide a continuous scale into intervals of uniform width.
For age 11–90, equal bins of width 10 years might look like:

11–20

20–30

30–40

...

80–90

# Example (Python using pd.cut):

```python
bins = list(range(10, 101, 10))
labels = [f"{bins[i]}-{bins[i+1]}" for i in range(len(bins)-1)]

df['Age_Bin'] = pd.cut(df['Age'], bins=bins, labels=labels, include_lowest=True)
print(df)
```

# Exploring pd.cut vs pd.qcut vs np.where
# a. pd.cut – Fixed Intervals

Used when you want predefined bins or equal-interval bins.

Example:

```python
pd.cut(df['Age'], bins=[10,20,30,40,50,60,70,80,90])
```

Use-case: grouping values into fixed numeric ranges.

# b. pd.qcut – Quantile-Based Binning

Bins contain approximately the same number of observations, rather than equal widths.

Example:

```python
pd.qcut(df['Age'], q=4)
```

This divides the dataset into quartiles (each bin holds 25% of the data).

Use-case: balanced segment sizes, useful in ML for stratification.

# np.where – Manual Conditional Grouping

Best when groups follow business logic or non-uniform boundaries.

Example:

```python
df['Category'] = np.where(df['Age'] < 20, 'Teen',
                   np.where(df['Age'] < 40, 'Adult',
                   np.where(df['Age'] < 60, 'Middle', 'Senior')))
```

Use-case: domain-driven classifications.

# Summary Table
| Method    | Type of Binning     | When to Use                                             |
|-----------|----------------------|----------------------------------------------------------|
| `pd.cut`  | Equal-width          | Analytical buckets, histograms, fixed intervals         |
| `pd.qcut` | Equal-frequency      | Balanced segmentation, quantiles, ML preprocessing      |
| `np.where`| Custom rules         | Human logic, business rules, non-uniform bins           |
