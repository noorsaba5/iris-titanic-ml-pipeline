```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Score": [70, 82, 90, 65]
})

df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.bar(df["Name"], df["Score"])
plt.title("Sample Scores")
plt.xlabel("Name")
plt.ylabel("Score")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = df[["Score"]]
y = [1, 2, 3, 4]   # just example levels

model = LinearRegression()
model.fit(X, y)

model.predict([[85]])
```
