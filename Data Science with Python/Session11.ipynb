```python
# pd.cut vs pd.qcut vs np.where
import pandas as pd
import numpy as np

stores = pd.read_csv('stores_new.csv')
stores.head()
```

```python
# pd.cut
stores['sales_cat'] = pd.cut(stores.TotalSales, bins = 4, labels=['standard', 'silver', 'gold', 'premium']) # Equally spaced intervals on the given numbers.
```

```python
stores.head()
```

```python
# pd.qcut

pd.qcut(stores.TotalSales, q = 4).value_counts(normalize=True) # All the 4 quantiles will have same data points (almost equal data points)
```

```python
# Capture the high vs low footfall (2 equally spaced intervals) and fetch total sales by these 2 categories.

stores['footfall'] = pd.cut(stores.TotalCustomers, bins = 2, labels=['low', 'high'])

stores.loc[stores.footfall == 'low', 'TotalSales'].sum()
```

```python
stores.loc[stores.footfall == 'high', 'TotalSales'].sum()
```

```python
stores.footfall.value_counts()
```

```python
# np.where() => similar to IF function on google sheets/excel
# Capture the high vs low footfall (> 150) and fetch total sales by these 2 categories.
stores['footfall'] = np.where(stores.TotalCustomers > 150, 'High', 'Low')
```

```python
stores.footfall.value_counts()
```

```python
stores.loc[stores.footfall == 'Low', 'TotalSales'].sum()
```

```python
stores.loc[stores.footfall == 'High', 'TotalSales'].sum()
```

```python
# Aggregations - sum, min, max, average, count

# group by
# pivot table
# pivot
# crosstab
```

```python
# What is the Location wise Sales?

stores.groupby('Location').TotalSales.sum()
```

```python
# What's the total # of stores per store type?
stores.groupby('StoreType').StoreCode.count()
```

```python
# What's the Location wise footfall and sales

stores.groupby('Location')[['TotalSales', 'TotalCustomers']].sum()
```

```python
stores.groupby('Location').agg({'TotalSales': 'sum', 'TotalCustomers': 'sum'}) # sum as a function => sum as a string
```

```python
# What's the Location wise average sales, total sales, min and max sales
stores.groupby('Location').TotalSales.agg(['sum', 'mean', 'min', 'max'])
```

```python
stores.groupby('Location').agg({'TotalSales': ['sum', 'mean', 'min', 'max']})
```

```python
# What's the total Sales by Location and Store Type

stores.groupby(['Location', 'StoreType']).TotalSales.sum().reset_index()
```

```python
stores.groupby(['Location', 'StoreType']).TotalSales.sum().reset_index().pivot(index = 'Location', columns = 'StoreType', values = 'TotalSales') #Cross table
```

```python
# Pivot tables -> Excel 
# Fetch Average total cost per store type and store name
# Hierarchical data
pd.pivot_table(data = stores, index = 'StoreName', columns='StoreType', values = 'TotalCost', aggfunc = 'mean', fill_value=0)
```

```python
# Fetch Average total cost per Location and store name
# Non-hierarchical data
pd.pivot_table(data = stores, index = 'StoreName', columns='Location', values = 'TotalCost', aggfunc = 'mean', fill_value=0)
```

```python
# Fetch total stores per location and StoreType

pd.pivot_table(data = stores, index = 'StoreType', columns='Location', values = 'StoreCode', aggfunc = 'count', fill_value=0)
```

```python
# pd.crosstab - Will provide counts across 2 different columns

pd.crosstab(index=stores.StoreType, columns=stores.Location)
```

```python
pd.merge(demog, trans, left_on= 'CustName', right_on='CustomerName', how = 'outer', indicator=True)
```

```python
# Identify the customers who never did any transactions

# left
temp = pd.merge(demog, trans, left_on= 'CustName', right_on='CustomerName', how = 'left', indicator=True)
temp[temp._merge == 'left_only']
```

```python
# What's the overall transactional value by each location?
# For location where we don't have customers, just put 0

temp = pd.merge(demog, trans, left_on= 'CustName', right_on='CustomerName', how = 'left')
temp.groupby('Location').Total_Transaction_value.sum()
```

```python
# Define average transactional value by location and Education in form of cross table.

pd.pivot_table(data = pd.merge(demog, trans, left_on= 'CustName', right_on='CustomerName', how = 'inner'),
                index = 'Location', columns = 'Education', aggfunc='mean', values = 'Total_Transaction_value', fill_value=0          
)
```
