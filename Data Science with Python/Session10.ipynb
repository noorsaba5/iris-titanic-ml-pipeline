```python
# UDF for percentile rule
def percentile(x): 
    # Calculate the outliers based on percentile rule - guarantees that highest and lowest 1 percentile is outlier
    lc = x.quantile(0.01) # np.quantile(x, 0.01)
    uc = x.quantile(0.99) # np.quantile(x, 0.99)

    return x.min(), lc, uc, x.max()

# This is only going to affect 2 percentile of data
percentile(stores.TotalSales)
```

```python
# UDF for iqr rule
def iqr(x):
    q1 = x.quantile(0.25)
    q3 = x.quantile(0.75)
    iqr = q3 - q1

    lc = q1 - 1.5 * iqr
    uc = q3 + 1.5 * iqr

    return x.min(), lc, uc, x.max()

iqr(stores.TotalSales)
```

```python
iqr(stores.TotalCustomers)
```

```python
# UDF for z-score rule
def z_score(x):

    mean = x.mean()
    std = x.std()

    lc = mean - 3 * std
    uc = mean + 3 * std
    return x.min(), lc, uc, x.max()

z_score(stores.TotalSales)
```
