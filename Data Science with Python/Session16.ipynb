```python
# Q6. Applying pearson's correlation

#H0: There is no relationship between latest mon usage and pre usage
#Ha: There is a relationship between latest mon usage and pre usage

alpha = 0.05

statistic,p_value = stats.pearsonr(df.pre_usage, df.Latest_mon_usage)
```

```python
if p_value < alpha:
    print('reject the null hypothesis; There is a relationship between latest mon usage and pre usage')
else:
    print('fail to reject the null hypothesis; There is no relationship between latest mon usage and pre usage')
```

```python
# Q4. ANOVA test
# H0: No difference in credit card usage of three different segments
# Ha: There is a difference in credit card usage of three different segments
```

```python
var = 'pre_usage'
seg1 = df.loc[df.segment == 1, var]
seg2 = df.loc[df.segment == 2, var]
seg3 = df.loc[df.segment == 3, var]
```

```python
statistic, p_value = stats.f_oneway(seg1, seg2, seg3)
```

```python
if p_value < alpha:
    print('reject the null hypothesis; There is a difference in credit card usage of three different segments')
else:
    print('fail to reject the null hypothesis; There is a difference in credit card usage of three different segments')
```
