```python
data = [2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7]

type_count = {}

for item in data:
    t = type(item).__name__
    type_count[t] = type_count.get(t, 0) + 1

print(type_count)
```

```python
data = [2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7]

total = 0

for item in data:
    if isinstance(item, (int, float, bool)):
        total += item

print("Sum of numeric values:", total)
```

```python
def my_func(l1):
    total_sum = 0
    str_count = 0
    
    for item in l1:
        # Sum only int and float (exclude bool, which is a subclass of int)
        if isinstance(item, (int, float)) and not isinstance(item, bool):
            total_sum += item
        
        # Count only strings
        if isinstance(item, str):
            str_count += 1
    
    return {"Sum": total_sum, "Count_of_Strs": str_count}


# Test
l1 = [2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7]
print(my_func(l1))
```

```python
def descriptives(*nums):
    # Filter only numeric values (int & float); exclude bool
    values = [n for n in nums if isinstance(n, (int, float)) and not isinstance(n, bool)]
    
    if not values:
        return {"Error": "No valid numeric inputs provided"}
    
    total = sum(values)
    count = len(values)
    minimum = min(values)
    maximum = max(values)
    mean = total / count
    
    return {
        "sum": total,
        "count": count,
        "min": minimum,
        "mean": mean,
        "max": maximum
    }


# Example
print(descriptives(10, 5, 2.5, 17, 3))
```

```python
def fn_area(shape, **kwargs):
    shape = shape.lower()

    # --- Square ---
    if shape == "square":
        if "side" not in kwargs:
            return "Please enter 'side' for a square"
        side = kwargs["side"]
        if not isinstance(side, (int, float)):
            return "Side must be numeric"
        return side ** 2

    # --- Rectangle ---
    elif shape == "rectangle":
        if "length" not in kwargs or "width" not in kwargs:
            return "Please enter both 'length' and 'width' for a rectangle"
        length = kwargs["length"]
        width = kwargs["width"]
        if not (isinstance(length, (int, float)) and isinstance(width, (int, float))):
            return "Length and width must be numeric"
        return length * width

    # --- Circle ---
    elif shape == "circle":
        if "radius" not in kwargs:
            return "Please enter 'radius' for a circle"
        radius = kwargs["radius"]
        if not isinstance(radius, (int, float)):
            return "Radius must be numeric"
        return 3.14159 * (radius ** 2)

    else:
        return "Unsupported shape. Choose from square, rectangle, or circle."


# Examples
print(fn_area("square", side=10))         # 100
print(fn_area("rectangle", length=5, width=4))   # 20
print(fn_area("circle", radius=7))        # 153.93791
```

```python
def func(l1, l2):
    # Convert to sets for efficient comparison
    set1 = set(l1)
    set2 = set(l2)

    matched = sorted(list(set1 & set2))
    only_l1 = sorted(list(set1 - set2))
    only_l2 = sorted(list(set2 - set1))

    return {
        "Matched": matched,
        "Only in l1": only_l1,
        "Only in l2": only_l2
    }


# Input lists
l1 = ['January', 'February', 'March', 'May', 'June', 'September', 'December']
l2 = ['January', 'February', 'April', 'June', 'October', 'December']

# Run function
print(func(l1, l2))
```
