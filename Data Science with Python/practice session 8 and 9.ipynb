```python
# Deriving calculated columns (creating new columns)

# Get the sales values after applying 5% of the discount on the products

#stores['Discounted Sales'] = stores.TotalSales - stores.TotalSales * 0.05 # Broadcasting + vectorization are allowed
```

```python
#Create a new column: Discounted Sales
#    - Apply a 5% discount (multiply by 0.95)

stores['Discounted Sales'] = stores['TotalSales'] * 0.95   # 5% discount applied
stores.head() # display first 5 rows of the dataframe
```

```python
stores.head()
```

```python
#Create a new column: Total Cost
#    Formula: OperatingCost + (Total_Customers * AcqCostPercust)

stores['TotalCost'] = (
    stores['OperatingCost'] 
    + stores['Total_Customers'] * stores['AcqCostPercust'] # Total customer acquisition cost
)

stores.head()
```

# Below 22 is a clean, easy-to-understand, fully commented version explaining .assign(), .insert(), and the concepts of *args and **kwargs in a practical, simple way.

```python
# -------------------------------------------------------------
# Using .assign() to create multiple new columns at once
# -------------------------------------------------------------

# .assign() lets you define several new columns in a single call.
# Each argument inside assign() becomes a new column in the dataframe.
# Syntax: .assign(NewColumn = expression, AnotherColumn = value)

stores = stores.assign(
    
    # Creating TotalCost2: same formula as before
    TotalCost2 = stores['OperatingCost'] 
                 + stores['Total_Customers'] * stores['AcqCostPercust'],
    
    # Creating RandomInfo: a constant value for demonstration
    RandomInfo = 10
)

stores.head()
```

```python
stores.head()
```

```python
# -------------------------------------------------------------
# Dropping multiple columns from the dataframe
# -------------------------------------------------------------

# .drop() removes columns or rows from a dataframe.
# columns = [...] → list of column names you want to remove
# inplace=True → updates the existing dataframe directly (no need to reassign)

stores.drop(
    columns=['TotalCost2', 'RandomInfo', 'TotalCost3'], 
    inplace=True
)

# Display the updated dataframe to confirm the columns were removed
stores.head()
```

```python
# -------------------------------------------------------------
# Filtering the dataframe to get specific rows
# -------------------------------------------------------------

# Goal: Fetch only the stores that are located in Chennai.
# We use a boolean condition: stores.Location == 'Chennai'
# This condition returns True for rows where the location is Chennai.

chennai_stores = stores[stores['Location'] == 'Chennai']

# Display the filtered result
chennai_stores
```

```python
# -------------------------------------------------------------
# Filtering the dataframe for multiple locations
# -------------------------------------------------------------

# Goal: Select stores that are located in either Chennai OR Delhi.
# We use the OR operator | to combine two conditions.
# Each condition checks the Location column for a specific city.

selected_stores = stores[
    (stores['Location'] == 'Chennai') |
    (stores['Location'] == 'Delhi')
]

# Display the filtered rows
selected_stores
```

```python
# Select only the stores located in Chennai or Delhi
filtered_stores = stores[stores['Location'].isin(['Chennai', 'Delhi'])]

# Display the result
print(filtered_stores)
```

# Step-by-step explanation:28

stores['Location'] → looks at the Location column in your DataFrame.

.isin(['Chennai', 'Delhi']) → checks each row to see if the location is either 'Chennai' or 'Delhi'.

stores[...] → keeps only the rows where the condition is True.

filtered_stores → now contains only the stores in Chennai or Delhi.

```python
# Get TotalSales and Total_Customers for stores in Chennai or Delhi
filtered_data = stores.loc[
    stores['Location'].isin(['Chennai', 'Delhi']),  # Keep rows where Location is Chennai or Delhi
    ['TotalSales', 'Total_Customers']              # Select only these two columns
]

print(filtered_data)
```

# Explanation in simple terms:29 

"Location in ['Chennai', 'Delhi']" is like saying: “Give me rows where Location is Chennai or Delhi”.

[['TotalSales', 'Total_Customers']] then picks only the columns you want.

Everything is compact and very readable, especially when dealing with multiple conditions.

This approach is great when you want the code to be almost like a natural-language filter.

```python
# Find stores where TotalCost is higher than the average TotalCost

# Step 1: Calculate the average TotalCost
average_total_cost = stores['TotalCost'].mean()

# Step 2: Filter stores with TotalCost greater than average
high_cost_stores = stores[stores['TotalCost'] > average_total_cost]

# Step 3: Display the result
print(high_cost_stores)
```

# Explanation in simple terms:30

stores['TotalCost'].mean() → calculates the average total cost across all stores.

stores['TotalCost'] > average_total_cost → creates a Boolean mask for stores above the average.

stores[...] → selects only the rows where the condition is True.

high_cost_stores → contains only stores with higher-than-average total cost.

This version is easier to read because it separates calculation from filtering, making it clear what each step does.
