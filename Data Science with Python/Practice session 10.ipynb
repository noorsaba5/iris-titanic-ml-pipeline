# Methods to Identify Outliers
# Percentile Rule (or Min-Max Threshold)

Based on percentiles of the data.

Commonly, anything below the 1st percentile or above the 99th percentile is considered an outlier.

```python
def percentile(x):
    #caculate outlier based on percentiles
    lc = x.quantile(0.01)
    uc = x.quantile(0.99)
    return x.min(), lc, uc, x.max()
#this is only going to affect 2 percentiles of the data
percentile(stores['TotalSales'])
```

```python


# Example: Store TotalSales
Q1 = stores['TotalSales'].quantile(0.01)
Q99 = stores['TotalSales'].quantile(0.99)

outliers_percentile = stores[(stores['TotalSales'] < Q1) | (stores['TotalSales'] > Q99)]
print(outliers_percentile)
```

# IQR Rule (Interquartile Range)

IQR = Q3 – Q1 (difference between 75th and 25th percentiles).

Anything below Q1 – 1.5×IQR or above Q3 + 1.5×IQR is an outlier.

```python
Q1 = stores['TotalSales'].quantile(0.25)
Q3 = stores['TotalSales'].quantile(0.75)
IQR = Q3 - Q1

outliers_iqr = stores[(stores['TotalSales'] < (Q1 - 1.5 * IQR)) |
                      (stores['TotalSales'] > (Q3 + 1.5 * IQR))]
print(outliers_iqr)
```

# Z-Score Rule

Measures how many standard deviations a value is from the mean.

Common threshold: |z| > 3 is considered an outlier.

✅ Note: Z-score works best for normally distributed data.

```python
from scipy import stats
import numpy as np

z_scores = np.abs(stats.zscore(stores['TotalSales']))
outliers_z = stores[z_scores > 3]
print(outliers_z)
```

# Keep Outliers

Sometimes outliers are important (e.g., extremely profitable stores) and should be analyzed separately.

| Method          | How it Works                          | When to Use                             |
| --------------- | ------------------------------------ | -------------------------------------- |
| Percentile Rule | Below 1st or above 99th percentile   | General, non-parametric datasets       |
| IQR Rule        | Below Q1–1.5*IQR or above Q3+1.5*IQR | Skewed or non-normal data              |
| Z-Score Rule    | |z| > 3 standard deviations from mean | Normally distributed data               |
| Handling        | Remove, cap, transform, or keep      | Depends on analysis and business context |


```python
def detect_outliers_iqr (df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[column] < (Q1 - 1.5 * IQR)) | (df[column] > (Q3 + 1.5 * IQR))]
    return outliers
```

```python
def detect_outliers_iqr(df, column):
    """
    Detects outliers in a DataFrame column using the IQR rule.
    
    Parameters:
    df (pd.DataFrame): The DataFrame containing the data
    column (str): The column name to check for outliers
    
    Returns:
    pd.DataFrame: DataFrame containing only the outlier rows
    """
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    # Identify outliers
    outliers = df[(df[column] < (Q1 - 1.5 * IQR)) | (df[column] > (Q3 + 1.5 * IQR))]

    return outliers

# -------------------------------
# Example Usage
# -------------------------------
# outliers_total_sales = detect_outliers_iqr(stores, 'TotalSales')
# print(outliers_total_sales)
```

```python
# UDF for iqr rule

def iqr(x): 

    q1 = x.quantile(0.25)

    q3 = x.quantile(0.75)

    iqr = q3 - q1

    lc = q1 - 1.5 * iqr

    uc = q3 + 1.5 * iqr

    return x.min(), lc, uc, x.max()

iqr(stores.TotalSales)
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# -------------------------------
# UDF for Z-Score Rule
# -------------------------------
def zscore_bounds(x, threshold=3):
    """
    Calculates the Z-Score bounds for outliers in a numeric series.
    
    Parameters:
    x (pd.Series): Numeric column
    threshold (float): Z-score threshold to define outliers (default = 3)
    
    Returns:
    tuple: (min_value, lower_bound, upper_bound, max_value)
           Values outside lower/upper bound are considered outliers
    """
    mean = x.mean()            # Mean of the column
    std = x.std()              # Standard deviation of the column
    
    lower_bound = mean - threshold * std
    upper_bound = mean + threshold * std
    
    return x.min(), lower_bound, upper_bound, x.max()

# -------------------------------
# Example Usage
# -------------------------------
zscore_bounds(stores['TotalSales'])
```

# Key points:

np.clip() does not print anything; it returns an array with the clipped values.

Assign it to a new column (stores['TotalSales_clipped']) to save the results.

Use print() or head() to display the values you want to see.

```python
Q1 = stores['TotalSales'].quantile(0.25)
Q3 = stores['TotalSales'].quantile(0.75)
IQR = Q3 - Q1

# Clip the TotalSales column to the IQR bounds
stores['TotalSales_clipped'] = np.clip(
    stores['TotalSales'], 
    Q1 - 1.5 * IQR,  # lower bound
    Q3 + 1.5 * IQR   # upper bound
)
# Display the first few rows to see the new column
stores[['TotalSales', 'TotalSales_clipped']].head()
```
