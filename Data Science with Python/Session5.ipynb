### List comprehension - List in one line.

```python

# [item if condition/s else item2 for item in iterable if condition/s]
# First if condition/s -> modification of the data
# Second if condition/s -> filtering of the data

lst1 = [var+100 if var>5 else var+200 for var in range(1,11) if var%2 == 0]
lst1
```

```python
lst = []
for i in range(1,11):
    if i%2 == 0:
        if i > 5:
            lst.append(i+100)
        else:
            lst.append(i+200)

lst
i
```

```python
# OOPS concepts - Object oriented programming

# class - Template
    # Attributes/Properties - variables in the class
    # Methods - Function within the class 
# object - Child of a class

class student:
    # name and grad exist within the scope of the class
    # self provides an instance of the current class
    # get_name, get_grad, set_name, set_grad are method of the class defined within the scope of the class
    def __init__(self, name, grad): # Constructor within the student class
        self.name = name
        self.grad = grad

    def get_name(self): # Fetch the name of current student
        return self.name
    
    def get_grad(self): # Fetch the grad of current student
        return self.grad

    def set_name(self, name):
        self.name = name

    def set_grad(self, grad):
        self.grad = grad
```

```python
def sum_nums(a, b = 0, c = 0): 
    """
        Input (Parameters):
            a: int, float
            b: int, float
        Sum of given values
    """
    if type(a) in [int, float] and type(b) in [int, float] and type(c) in [int, float]:
        return a + b + c
    else:
        print('Provide correct values!!!')
    return 0

sum_nums(c = 4, b = 2, 1) # Define the arguments
```

```python
# Given n numbers, identify their sum.
# 1 + 2 + 3 + 4 = 10
# 1 + 2 = 3

# *args - will capture n items in a tuple within the function

def sum_elements(*args):
    total = 0
    for item in args:
        if type(item) not in [int, float]:
            pass
        else:
            total = total + item

    # return sum([item for item in args if type(item) in [int, float]])
    return total

sum_elements(1,2,3,4, 'xyz', 1)
```

```python
# Perform arithmetic operation given by user on n numbers

# *args - will capture n items in a tuple within the function

def arithmetic_ops(*args, op = 'sum'): # Optional (keyword) arg + *args for handling multiple values
    total = 0 if op == 'sum' else 1
    print(args)
    for item in args:
        if type(item) not in [int, float]:
            pass
        elif op == 'sum':
            total = total + item
        elif op == 'mul':
            total = total * item
        else:
            print('invalid operation')

    # return sum([item for item in args if type(item) in [int, float]])
    return total

arithmetic_ops(1,2,3,4, 'xyz', 1)
```

```python
# Write a UDF that identify the duplicates within the 2 lists having n numbers.
# Avoid using set
# l1 = [1,2,3,4]
# l1 = [3,4,5]
# => [3,4]
# Do proper data validation

def identify_dups(l1:list = [], l2:list = []):
    
    if len(l1) == 0 or len(l2) == 0:
        print('no dups')

    for item in l1:
        if type(item) not in [int, float]:
            return 'invalid value in list'
        
    for item in l2:
        if type(item) not in [int, float]:
            return 'invalid value in list'
    
    l3 = []
    # O(nm) - Try dictionary?
    for item in l1:
        if item in l2: # Time complexity? - O(n)
            l3.append(item)

    return l3

identify_dups([1,2], [2,4])
```
