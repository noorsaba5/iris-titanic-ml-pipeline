```python
# Q. Identify if the given input is even or odd
# Handle the input validation - Error handling
# try & except block will be used whenever we anticipate that a code block can throw an error.
try:
    n = int(float(input()))

    print(n % 2 == 0) # Even
    print(n % 2 == 1) # Odd
except:
    print("Please provide valid input in numbers!!!")

print("Exit!!!")
```

### 1.3 Control Flow statements

```python
# Define if the given number is positive or negative or 0.
n = 0

if n > 0:
    print('positive')
elif n == 0:
    print(0)
else:
    print('negative')
```

```python
# Consider that we have a user input, and we have to print
# 1. 'foo' if the number is completely divisible by 3
# 2. 'bar' if the number is completely divisible by 5
# 3. 'foobar' if the number is compeltely divisibly by both 3 and 5
# 4. print the number if none is the case.

n = int(input())

if n % 3 == 0 and n % 5 == 0:
    print('foobar')
elif n % 3 == 0:
    print('foo')
elif n % 5 == 0:
    print('bar')
else:
    print(n)
```
