```python
# Within a set of numbers from 1 to 100, identify the even and odd numbers and add them in their individual list.

even = []
odd = []

for item in range(1, 101):
    if item % 2 == 0:
        even.append(item)
    else:
        odd.append(item)
print(odd)
```

```python
# Q. Given a list of numbers, count all the occurrences of the numbers that exist.
# Range of numbers - 1 to 100
lst = [2,3,4,5,77,99,3,4,6,7,89,77,77,66,55]
"""
2 - 1
3 - 2
4 - 2
...
"""
result = [0] * max(lst)

for item in lst:
    if item > 100 or item < 0:
        continue
    result[item-1] += 1

for i in range(len(result)): # enumerate(result) - indexes along with individual item
    if result[i] > 0:
        print(i+1, result[i])
```

```python
# Count the elements of each datatype inside the list and display in output
lst = [2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7]   

int_n = 0
float_n = 0
str_n = 0
bool_n = 0
other_n = 0
for item in lst:
    if type(item) == int:
        int_n+=1
    elif type(item) == float:
        float_n+=1
    elif type(item) == str:
        str_n+=1
    elif type(item) == bool:
        bool_n+=1
    else:
        other_n+=1

print(int_n, float_n, str_n, bool_n, other_n)
```

```python
# Given a list of numbers, count all the occurrences of the numbers that exist.
# Range of numbers - 1 to 100

lst = [2,3,4,5,77,99,3,4,6,7,89,77,77,66,55]

d = dict()

for item in lst:
    if item in d: #This is met when item is already past of the dictionary
        d[item]+=1
    else: # This is met when item is not part of the dictionary
        d[item]=1

print(d)
```

```python
# Count the elements of each datatype inside the list and display in output
lst = [2, 3, 'Py', '10', 1, 'SQL', 5.5, True, 3, 'John', None, 7]

d = dict()

for item in lst:
    if type(item) in d: #This is met when item is already past of the dictionary
        d[type(item)]+=1
    else: # This is met when item is not part of the dictionary
        d[type(item)]=1

d
```
